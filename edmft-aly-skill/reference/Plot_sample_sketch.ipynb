### Step 1 - Reload the `datacol` + build the album

`create_page` takes a **string** tag - one of your `$PLOT_TARGETS`. Each page's `init()`
builds the overview `sketch` plus per-manifold figures in `fig_list`.

```python
col = data.load_dcl('<CASE>_data.pkl')
alb = qplt.album()
for tag in ['sig', 'cdos', 'gc1', 'dlt1']:   # trim to your $PLOT_TARGETS
    alb.create_page(col.mission_list, tag)
print(alb)                                   # lists pages: index <-> key_tag
```
