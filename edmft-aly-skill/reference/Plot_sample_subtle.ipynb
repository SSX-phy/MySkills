## Stage 3 - Plot -> figures + a pickled `album`

Sample for `edmft-aly-skill` Section 3. Copy this into `$PLOT_WORKDIR` as `Plot.ipynb`.
It reloads the `datacol` pickled in Stage 2, builds an `album`, renders the requested
figures, and pickles the album (with its live figures) so you can re-open and tweak.

**Comment-out-when-done:** once a figure looks right, wrap its build/tweak code in a
`'''...'''` block and leave the cell as a single display line. To adjust later,
un-comment that block, edit, re-run.

### Step 0 - Imports + backend

Remote (`$PLOT_RUN_LOC = remote`, default): use the `Agg` backend and `savefig` to PDF,
then download the PDFs. Local: use `%matplotlib inline` instead of the `Agg` line.

```python
import matplotlib
matplotlib.use('Agg')        # remote default; locally use:  %matplotlib inline
import data
import quickplot as qplt
```

### Step 1 - Reload the `datacol` + build the album

`create_page` takes a **string** tag - one of your `$PLOT_TARGETS`
(`'sig'`,`'cdos'`,`'gc1'`,`'dlt1'`). Each page's `init()` builds an overview `sketch`
figure plus per-manifold figures in `fig_list`.

```python
col = data.load_dcl('<CASE>_data.pkl')
alb = qplt.album()
for tag in ['sig', 'cdos', 'gc1', 'dlt1']:   # trim to your $PLOT_TARGETS
    alb.create_page(col.mission_list, tag)
print(alb)                                   # lists pages: index <-> key_tag
```

#### What each target shows
- **sig** (`GF/sig.inp` = real-axis Sigma): Re Sigma - Sigma(inf) (in-gap pole => Mott) and -Im Sigma.
- **cdos**: col1 total DOS, col2 partial correlated-shell DOS; gap / QP weight at E_F.
- **gc1**: -Im G per manifold (orbital-resolved DOS).
- **dlt1**: -Im Delta(omega), the hybridization.
